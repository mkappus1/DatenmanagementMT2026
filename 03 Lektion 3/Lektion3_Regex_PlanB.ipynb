## Schritt 2: Parameter anpassen

**Nur diese Zelle ändern Sie** – genau wie den oberen Teil von `regex_suche.py`:

| Parameter | Bedeutung |
|---|---|
| `DATEI` | Welche Datei wird durchsucht? |
| `MUSTER` | Der reguläre Ausdruck. Das `r` vor den Anführungszeichen bitte stehen lassen! |
| `MODUS` | `"zeilen"` = ganze Zeilen mit Treffer · `"treffer"` = nur die gefundenen Stellen · `"ersetzen"` = MUSTER durch ERSATZ ersetzen |
| `ERSATZ` | Nur für den Modus `"ersetzen"` |
| `AUSGABE` | Name der Ergebnisdatei. Beim Ersetzen **nie** den Namen der Originaldatei verwenden! |

Tipp: Enthält Ihr Muster selbst doppelte Anführungszeichen (z. B. `xml:lang="de"`), setzen Sie es in einfache: `r'<tuv xml:lang="de">...'`

Nach dem Ändern die Zelle mit ▶ starten.

````python
# ---- Parameter anpassen ----
DATEI   = "beispiel.tmx"
MUSTER  = r"Climate"
MODUS   = "zeilen"      # "zeilen", "treffer" oder "ersetzen"
ERSATZ  = ""            # nur für MODUS "ersetzen"
AUSGABE = "climate.txt"
# ----------------------------
print("Parameter gespeichert. Weiter mit Schritt 3.")
````

## Schritt 3: Suche ausführen

Diese Zelle **nicht ändern**, nur starten ▶. Sie zeigt die Anzahl Treffer und die ersten Ergebnisse an und speichert alles in der AUSGABE-Datei.

````python
import re

with open(DATEI, encoding="utf-8") as f:
    text = f.read()

if MODUS == "zeilen":
    ergebnis = [z for z in text.splitlines() if re.search(MUSTER, z)]
elif MODUS == "treffer":
    ergebnis = re.findall(MUSTER, text)
elif MODUS == "ersetzen":
    neu, anzahl = re.subn(MUSTER, ERSATZ, text)
    ergebnis = [neu]
else:
    raise ValueError('MODUS muss "zeilen", "treffer" oder "ersetzen" sein.')

with open(AUSGABE, "w", encoding="utf-8") as f:
    f.write("\n".join(ergebnis))

if MODUS == "ersetzen":
    print(anzahl, "Ersetzungen gespeichert in", AUSGABE)
else:
    print(len(ergebnis), "Treffer gespeichert in", AUSGABE)
    print("\nDie ersten Ergebnisse:")
    for e in ergebnis[:10]:
        print("  ", e)
````
